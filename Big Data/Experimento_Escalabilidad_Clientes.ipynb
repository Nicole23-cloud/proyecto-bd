```python
%sql
DESCRIBE workspace.olist_nuevo.limpio_clientes;
```

```python
%sql
SELECT c.customer_state, COUNT(*) AS clientes
FROM workspace.olist_nuevo.limpio_clientes c
GROUP BY c.customer_state
ORDER BY clientes DESC;
```

```python
%sql
SELECT COUNT(*) AS filas FROM workspace.olist_nuevo.limpio_clientes;
```

```python
%sql
CREATE OR REPLACE TABLE workspace.olist_nuevo.exp_clientes_x2 AS
SELECT * FROM workspace.olist_nuevo.limpio_clientes
UNION ALL
SELECT concat(customer_id,'_dup'), concat(customer_unique_id,'_dup'),
       customer_zip_code_prefix, customer_city, customer_state
FROM workspace.olist_nuevo.limpio_clientes;
```

```python
%sql
SELECT COUNT(*) AS filas, COUNT(DISTINCT customer_id) AS ids_unicos
FROM workspace.olist_nuevo.exp_clientes_x2;
```

```python
%sql
SELECT c.customer_state, COUNT(*) AS clientes
FROM workspace.olist_nuevo.exp_clientes_x2 c
GROUP BY c.customer_state
ORDER BY clientes DESC;
```
