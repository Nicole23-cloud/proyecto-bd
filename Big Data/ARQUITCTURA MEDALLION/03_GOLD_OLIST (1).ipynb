```python
%sql

CREATE OR REPLACE TEMP VIEW gold_ventas AS

SELECT
    d.order_id,
    d.order_item_id,

    -- CLIENTE
    c.customer_id,
    c.customer_unique_id,
    c.nombre_cliente,
    c.customer_city,
    c.customer_state,

    -- PEDIDO
    p.order_status,
    p.order_purchase_timestamp,

    -- PRODUCTO
    pr.product_id,
    pr.product_category_name,
    cat.product_category_name_english,

    -- VENDEDOR
    v.seller_id,
    v.nombre_vendedor,
    v.seller_city,
    v.seller_state,

    -- VALORES DE VENTA
    d.price,
    d.freight_value

FROM silver_detalles d

INNER JOIN silver_pedidos p
    ON d.order_id = p.order_id

INNER JOIN silver_clientes c
    ON p.customer_id = c.customer_id

INNER JOIN silver_productos pr
    ON d.product_id = pr.product_id

INNER JOIN silver_vendedores v
    ON d.seller_id = v.seller_id

INNER JOIN silver_categorias cat
    ON pr.product_category_name = cat.product_category_name

WHERE cat.product_category_name_english IS NOT NULL;
```

```python
%sql

CREATE OR REPLACE TABLE default.gold_kpi_02_top_vendedores AS

WITH ingresos_vendedor AS (
    SELECT
        v.seller_id,
        v.nombre_vendedor,
        ROUND(SUM(d.price), 2) AS ingresos_totales
    FROM silver_detalles d
    INNER JOIN silver_pedidos p
        ON d.order_id = p.order_id
    INNER JOIN silver_productos pr
        ON d.product_id = pr.product_id
    INNER JOIN silver_categorias cat
        ON pr.product_category_name = cat.product_category_name
    INNER JOIN silver_vendedores v
        ON d.seller_id = v.seller_id
    WHERE cat.product_category_name_english IS NOT NULL
    GROUP BY
        v.seller_id,
        v.nombre_vendedor
),
ranking AS (
    SELECT
        ROW_NUMBER() OVER (
            ORDER BY ingresos_totales DESC
        ) AS position,
        seller_id,
        nombre_vendedor,
        ingresos_totales
    FROM ingresos_vendedor
)
SELECT
    position,
    CONCAT(
        LPAD(CAST(position AS STRING), 2, '0'),
        '. ',
        nombre_vendedor
    ) AS nombre_vendedor,
    ingresos_totales
FROM ranking
WHERE position <= 10
ORDER BY position;

SELECT *
FROM default.gold_kpi_02_top_vendedores;
```

```python
%sql

CREATE OR REPLACE TABLE default.gold_kpi_03_ingresos_categoria AS

WITH ingresos_categoria AS (
    SELECT
        cat.product_category_name_english,
        ROUND(SUM(d.price), 2) AS ingresos_totales

    FROM silver_detalles d

    INNER JOIN silver_pedidos p
        ON d.order_id = p.order_id

    INNER JOIN silver_productos pr
        ON d.product_id = pr.product_id

    INNER JOIN silver_categorias cat
        ON pr.product_category_name = cat.product_category_name

    WHERE cat.product_category_name_english IS NOT NULL

    GROUP BY cat.product_category_name_english
),

ranking AS (
    SELECT
        ROW_NUMBER() OVER (
            ORDER BY ingresos_totales DESC
        ) AS position,
        product_category_name_english,
        ingresos_totales

    FROM ingresos_categoria
)

SELECT
    position,
    CONCAT(
        LPAD(CAST(position AS STRING), 2, '0'),
        '. ',
        product_category_name_english
    ) AS categoria_producto,
    ingresos_totales

FROM ranking
WHERE position <= 10
ORDER BY position;


SELECT *
FROM default.gold_kpi_03_ingresos_categoria;
```

```python
%sql

CREATE OR REPLACE TABLE default.gold_kpi_04_evolucion_mensual AS

SELECT
    YEAR(p.order_purchase_timestamp) AS anio,

    MONTH(p.order_purchase_timestamp) AS numero_mes,

    CASE MONTH(p.order_purchase_timestamp)
        WHEN 1 THEN '01 Ene'
        WHEN 2 THEN '02 Feb'
        WHEN 3 THEN '03 Mar'
        WHEN 4 THEN '04 Abr'
        WHEN 5 THEN '05 May'
        WHEN 6 THEN '06 Jun'
        WHEN 7 THEN '07 Jul'
        WHEN 8 THEN '08 Ago'
        WHEN 9 THEN '09 Sep'
        WHEN 10 THEN '10 Oct'
        WHEN 11 THEN '11 Nov'
        WHEN 12 THEN '12 Dic'
    END AS nombre_mes,

    ROUND(SUM(d.price), 2) AS ingresos_totales

FROM silver_detalles d

INNER JOIN silver_pedidos p
    ON d.order_id = p.order_id

INNER JOIN silver_productos pr
    ON d.product_id = pr.product_id

INNER JOIN silver_categorias cat
    ON pr.product_category_name = cat.product_category_name

WHERE cat.product_category_name_english IS NOT NULL

GROUP BY
    YEAR(p.order_purchase_timestamp),
    MONTH(p.order_purchase_timestamp)

ORDER BY
    anio ASC,
    numero_mes ASC;

SELECT *
FROM default.gold_kpi_04_evolucion_mensual
ORDER BY anio ASC, numero_mes ASC;
```

```python
%sql

CREATE OR REPLACE TABLE default.gold_kpi_05_variacion_mensual AS

WITH ingresos_mensuales AS (
    SELECT
        YEAR(p.order_purchase_timestamp) AS anio,
        MONTH(p.order_purchase_timestamp) AS mes,
        ROUND(SUM(d.price), 2) AS ingresos

    FROM silver_detalles d

    INNER JOIN silver_pedidos p
        ON d.order_id = p.order_id

    INNER JOIN silver_productos pr
        ON d.product_id = pr.product_id

    INNER JOIN silver_categorias cat
        ON pr.product_category_name = cat.product_category_name

    WHERE cat.product_category_name_english IS NOT NULL

    GROUP BY
        YEAR(p.order_purchase_timestamp),
        MONTH(p.order_purchase_timestamp)
),

comparacion AS (
    SELECT
        actual.anio,
        actual.mes,
        actual.ingresos AS ingresos_actuales,
        anterior.ingresos AS ingresos_anteriores

    FROM ingresos_mensuales actual

    LEFT JOIN ingresos_mensuales anterior
        ON anterior.anio = CASE
            WHEN actual.mes = 1 THEN actual.anio - 1
            ELSE actual.anio
        END

        AND anterior.mes = CASE
            WHEN actual.mes = 1 THEN 12
            ELSE actual.mes - 1
        END
)

SELECT
    ROW_NUMBER() OVER (
        ORDER BY anio, mes
    ) AS position,

    anio,
    mes,

    CONCAT(
        LPAD(CAST(mes AS STRING), 2, '0'),
        '. ',
        CASE mes
            WHEN 1 THEN 'Ene'
            WHEN 2 THEN 'Feb'
            WHEN 3 THEN 'Mar'
            WHEN 4 THEN 'Abr'
            WHEN 5 THEN 'May'
            WHEN 6 THEN 'Jun'
            WHEN 7 THEN 'Jul'
            WHEN 8 THEN 'Ago'
            WHEN 9 THEN 'Sep'
            WHEN 10 THEN 'Oct'
            WHEN 11 THEN 'Nov'
            WHEN 12 THEN 'Dic'
        END,
        ' ',
        CAST(anio AS STRING)
    ) AS periodo,

    ROUND(ingresos_actuales, 2) AS ingresos_actuales,

    ROUND(ingresos_anteriores, 2) AS ingresos_anteriores,

    ROUND(
        ingresos_actuales - ingresos_anteriores, 2
    ) AS diferencia_ingresos,

    ROUND(
        (ingresos_actuales - ingresos_anteriores)
        / NULLIF(ingresos_anteriores, 0) * 100,
        2
    ) AS variacion_porcentual

FROM comparacion

ORDER BY anio, mes;

SELECT *
FROM default.gold_kpi_05_variacion_mensual
ORDER BY position;
```

```python
%sql

CREATE OR REPLACE TABLE default.gold_kpi_06_distribucion_geografica AS

WITH coords (state, latitud, longitud) AS (
    VALUES
    ('AC', -9.97, -67.81), ('AL', -9.67, -35.74), ('AP', 0.03, -51.07),
    ('AM', -3.12, -60.02), ('BA', -12.97, -38.50), ('CE', -3.73, -38.52),
    ('DF', -15.79, -47.88), ('ES', -20.32, -40.34), ('GO', -16.68, -49.25),
    ('MA', -2.53, -44.30), ('MT', -15.60, -56.10), ('MS', -20.47, -54.62),
    ('MG', -19.92, -43.94), ('PA', -1.46, -48.50), ('PB', -7.12, -34.86),
    ('PR', -25.43, -49.27), ('PE', -8.05, -34.88), ('PI', -5.09, -42.80),
    ('RJ', -22.91, -43.17), ('RN', -5.79, -35.21), ('RS', -30.03, -51.23),
    ('RO', -8.76, -63.90), ('RR', 2.82, -60.67), ('SC', -27.60, -48.55),
    ('SP', -23.55, -46.63), ('SE', -10.91, -37.07), ('TO', -10.18, -48.33)
),

ventas_estado AS (
    SELECT
        c.customer_state AS state,
        ROUND(SUM(d.price), 2) AS ingresos_totales

    FROM silver_detalles d

    INNER JOIN silver_pedidos p
        ON d.order_id = p.order_id

    INNER JOIN silver_clientes c
        ON p.customer_id = c.customer_id

    INNER JOIN silver_productos pr
        ON d.product_id = pr.product_id

    INNER JOIN silver_categorias cat
        ON pr.product_category_name = cat.product_category_name

    WHERE cat.product_category_name_english IS NOT NULL

    GROUP BY c.customer_state
)

SELECT
    v.state,
    c.latitud,
    c.longitud,
    v.ingresos_totales

FROM ventas_estado v

INNER JOIN coords c
    ON v.state = c.state;

SELECT *
FROM default.gold_kpi_06_distribucion_geografica;
```

```python
%sql
WITH ingresos_estado AS (
    SELECT
        customer_state AS estado,
        SUM(price) AS ingresos_totales
    FROM ef.ventas
    GROUP BY customer_state
),
ranking AS (
    SELECT
        ROW_NUMBER() OVER (
            ORDER BY ingresos_totales DESC
        ) AS posicion,
        estado,
        ingresos_totales
    FROM ingresos_estado
)
SELECT
    CONCAT(
        LPAD(CAST(posicion AS STRING), 2, '0'),
        '. ',
        estado
    ) AS estado,
    ROUND(ingresos_totales, 2) AS ingresos,
    ROUND(
        ingresos_totales * 100.0 / SUM(ingresos_totales) OVER (),
        2
    ) AS porcentaje_ventas
FROM ranking
WHERE posicion <= 10
ORDER BY posicion ASC;
```
